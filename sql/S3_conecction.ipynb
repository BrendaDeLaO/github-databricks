```python
# 1. Ruta de tu archivo en el S3
ruta = "s3://amankay/emergencias_danos_silver_2003_2025 (1).csv"

# 2. Lectura del CSV
df = (
    spark.read
    .option("header", "true")
    .option("delimiter", ",")      # cámbialo a "," si usa coma
    .option("inferSchema", "true")
    .option("encoding", "UTF-8")
    .csv(ruta)
)

# 3. Guardar en el catálogo
nombre_tabla = "big_data_team1.brenda.datos_indeci"

df.write.mode("overwrite").saveAsTable(nombre_tabla)

print(f"¡Listo! Tabla guardada con éxito en: {nombre_tabla}")
```
