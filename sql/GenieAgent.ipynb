```python
%sql
---5 departamentos con el mayor número total de fallecidos, heridos y personas afectadas registrados a lo largo de todo el periodo que abarca el conjunto de datos---
SELECT
    `dpto`,
    SUM(`fallecidos`) AS total_fallecidos,
    SUM(`lesionados`) AS total_lesionados,
    SUM(`afectados`) AS total_afectados,
    SUM(`fallecidos` + `lesionados` + `afectados`) AS total_casualties
FROM `big_data_team1`.`brenda`.`datos_indeci`
WHERE `dpto` IS NOT NULL
GROUP BY `dpto`
ORDER BY total_casualties DESC
LIMIT 5;
```

```python
%sql
---comparar el costo de la emergencia (costo_de_la_emergencia) y la cantidad de emergencias registradas año por año (ano), ordenado del año con mayor costo acumulado al menor---
SELECT
    `ano`,
    COUNT(*) AS numero_de_emergencias,
    SUM(`costo_de_la_ayuda`) AS costo_total_ayuda,
    AVG(`costo_de_la_ayuda`) AS costo_promedio_ayuda
FROM `big_data_team1`.`brenda`.`datos_indeci`
WHERE `ano` IS NOT NULL
GROUP BY `ano`
ORDER BY costo_total_ayuda DESC
```

```python
%sql
---mostrar el número total de viviendas destruidas (*viviendas_destruidas*) y viviendas afectadas (*viviendas_afectadas*), específicamente para aquellos eventos en los que el tipo de peligro es 'SISMO' o 'VIENTOS FUERTES', agrupados por región natural (*region_natural*)---
SELECT
    `region_natural`,
    SUM(`viviendas_destruidas`) AS total_viviendas_destruidas,
    SUM(`viviendas_afectadas`) AS total_viviendas_afectadas,
    COUNT(*) AS numero_de_eventos
FROM `big_data_team1`.`brenda`.`datos_indeci`
WHERE `peligro` IN ('SISMO', 'VIENTOS FUERTES')
    AND `region_natural` IS NOT NULL
GROUP BY `region_natural`
ORDER BY total_viviendas_destruidas DESC
```

```python
%sql
---los 10 distritos (dist) más afectados en términos de personas afectadas (afectados) durante los últimos 5 años registrados---
SELECT
    `dist`,
    `dpto`,
    SUM(`afectados`) AS total_afectados,
    COUNT(*) AS numero_de_emergencias
FROM `big_data_team1`.`brenda`.`datos_indeci`
WHERE `ano` >= 2021
    AND `ano` <= 2025
    AND `dist` IS NOT NULL
GROUP BY `dist`, `dpto`
ORDER BY total_afectados DESC
LIMIT 10
```

```python
%sql
---calcular la suma total del costo de la emergencia (costo_de_la_emergencia) y el promedio del peso de la emergencia (peso_de_la_emergencia) agrupado por tipo de peligro (peligro), ordenando los resultados de mayor a menor costo total.---
SELECT
    `peligro`,
    SUM(`costo_de_la_ayuda`) AS costo_total,
    AVG(`peso_de_la_ayuda`) AS peso_promedio,
    COUNT(*) AS numero_de_emergencias
FROM `big_data_team1`.`brenda`.`datos_indeci`
WHERE `peligro` IS NOT NULL
GROUP BY `peligro`
ORDER BY costo_total DESC
```
